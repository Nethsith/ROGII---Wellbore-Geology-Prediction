```python
from pathlib import Path
import pandas as pd

def find_project_root():
    current = Path.cwd()

    for folder in [current, *current.parents]:
        if (folder / "data" / "raw").exists():
            return folder

    raise FileNotFoundError(
        f"Could not find project root containing data/raw.\n"
        f"Current folder: {current}"
    )

PROJECT_ROOT = find_project_root()
RAW_DIR = PROJECT_ROOT / "data" / "raw"
```

```python


results = []

for file in sorted(RAW_DIR.rglob("*_horizontal_well.csv")):
    df = pd.read_csv(file)

    results.append({
        "file": str(file),
        "rows": len(df),
        "X_missing": int(df["X"].isna().sum()),
        "Y_missing": int(df["Y"].isna().sum()),
        "Z_missing": int(df["Z"].isna().sum()),
    })

result_df = pd.DataFrame(results)

# Show only wells that actually have missing X, Y, or Z
missing_only = result_df[
    (result_df["X_missing"] > 0) |
    (result_df["Y_missing"] > 0) |
    (result_df["Z_missing"] > 0)
]

if missing_only.empty:
    print("Great — no missing X, Y, or Z values in any horizontal-well file.")
else:
    print(missing_only.to_string(index=False))
```

```python


results = []

for file in sorted(RAW_DIR.rglob("*.csv")):
    df = pd.read_csv(file)

    if "GR" not in df.columns:
        continue

    missing_idx = df.index[df["GR"].isna()].tolist()

    results.append({
        "file": str(file),
        "file_type": (
            "horizontal_well"
            if "__horizontal_well.csv" in file.name
            else "typewell"
        ),
        "rows": len(df),
        "GR_missing": len(missing_idx),
        "percentage": len(missing_idx)*100 /len(df),
        "first_missing_row": min(missing_idx) if missing_idx else None,
        "last_missing_row": max(missing_idx) if missing_idx else None,
    })

gr_check = pd.DataFrame(results)

missing_gr = gr_check[gr_check["GR_missing"] > 0]

if missing_gr.empty:
    print("Great — no missing GR values in any horizontal-well or typewell file.")
else:
    print(missing_gr.to_string(index=False))
```
